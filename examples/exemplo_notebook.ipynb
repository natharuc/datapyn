# Exemplo de Notebook Jupyter

Este notebook demonstra a importacao de notebooks no DataPyn

```python
import pandas as pd
import numpy as np
```

```python
# Criar DataFrame de exemplo
df = pd.DataFrame({
    'nome': ['Ana', 'Bruno', 'Carlos', 'Diana'],
    'idade': [25, 30, 35, 28],
    'cidade': ['Sao Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'Porto Alegre']
})
df
```

```python
# Calcular estatisticas
media_idade = df['idade'].mean()
print(f'Idade media: {media_idade:.1f} anos')
```

## Analise de Dados

Vamos filtrar os dados:

```python
# Filtrar pessoas com mais de 28 anos
df_filtrado = df[df['idade'] > 28]
df_filtrado
```
